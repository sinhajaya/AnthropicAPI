```python
#LOAD ENV VARIABLES
from dotenv import load_dotenv

load_dotenv()

#Create an API client
import os
from google import genai
from google.genai import types

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
model = "gemini-3.5-flash"
#noinspection PyTypeChecker
def add_user_message(messages,text):
    user_message = {"role": "user", "parts": [{"text": text}]}
    messages.append(user_message)

def add_assistant_message(messages,text):
    assistant_message = {"role": "model", "parts": [{"text": text}]}
    messages.append(assistant_message)

def chat(messages,system=None,temperature=1.0):

    params = {
        "model":model,
        "contents":messages,
        "config" : types.GenerateContentConfig(
            temperature=temperature
        )
    }
    if system:
        params["config"] = types.GenerateContentConfig(
            system_instruction=system
        )

    message = client.models.generate_content(**params)
    return message.text
```

```python
messages = []

add_user_message(messages,
                 "Generate a one sentence movie idea")
system = "You are Karan Johar a bollywood director.So adopt his style of films"
answer = chat(messages,system=system,temperature=1.0)
answer
```
