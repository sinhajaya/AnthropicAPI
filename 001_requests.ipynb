```python
#Create an API client
import os
from google import genai

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
model = "gemini-2.5-flash"
```

```python
#noinspection PyTypeChecker
def add_user_message(messages,text):
    user_message = {"role": "user", "parts": [{"text": text}]}
    messages.append(user_message)

def add_assistant_message(messages,text):
    assistant_message = {"role": "model", "parts": [{"text": text}]}
    messages.append(assistant_message)

def chat(messages):
    message = client.models.generate_content(
        model=model,
        contents=messages
    )
    return message.text
```

```python
#Make a starting list of messages
messages = []

#Add in the initial user question
add_user_message(messages, "Define quantum computing in one sentence")

#Pass the list of messages into 'chat' to get an answer
answer = chat(messages)

#Take the answer and add it to the assistant message in our list
add_assistant_message(messages, answer)

#Add in the user's follow-up questions
add_user_message(messages, "Write another sentence")

#Call chat again with the list of messages to get a final answer
answer = chat(messages)
answer
```
